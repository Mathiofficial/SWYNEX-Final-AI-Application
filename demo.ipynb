## 2. Error handling
Bad input returns a readable error instead of crashing.

```python
bad_inputs = ["", "     ", None, 12345, "help", "refund " * 600]
for b in bad_inputs:
    r = safe_triage(b, sentiment, classifier)
    print({k: (v if k != "input" else str(v)[:30]) for k, v in r.items()})
```
